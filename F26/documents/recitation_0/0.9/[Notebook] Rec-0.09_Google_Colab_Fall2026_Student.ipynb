# 4️⃣ GPU / TPU Runtime Setup

### How to Change Your Runtime
1. Go to **Runtime** → **Change runtime type**  
2. Set **Hardware accelerator** to **GPU** (or TPU)  
3. Click **Save**, then re-run the cell below  

```python
import torch

if torch.cuda.is_available():
    device = torch.device("cuda")
    print("✅ GPU is available!")
    print("   GPU name :", torch.cuda.get_device_name(0))
    print("   CUDA ver :", torch.version.cuda)
else:
    device = torch.device("cpu")
    print("⚠️" * 10)
    print("⚠️  NO GPU DETECTED — you are running on CPU!")
    print("⚠️  Go to  Runtime → Change runtime type → Select T4/A100 GPU.")
    print("⚠️" * 10)

print("\nActive device:", device)
```
